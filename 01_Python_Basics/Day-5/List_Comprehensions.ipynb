```python
#Create a list of squares of odd numbers from 1 to 20 using list comprehension
squares_of_odds = [x**2 for x in range(1, 21) if x % 2 != 0]
print(squares_of_odds)
```

```python
#Create a list of numbers from 1 to 20, but:

#If the number is even → store its square

#If the number is odd → store the number as-is

result = [x**2 if x % 2 == 0 else x for x in range(1, 21)]
print(result)
```

```python
#Create a new list that contains only numbers divisible by both 5 and 10.
divisible_by_both = [x for x in range(1, 101) if x % 5 == 0 and x % 10 == 0]
print(divisible_by_both)
```
