```python
import pandas as pd

data = [10, 20, 30, 40]
s = pd.Series(data)

print(s)
```

```python
s = pd.Series([90, 85, 88], index=["Math", "Science", "English"])
print(s)

print(s["Math"])      # by label
print(s[0])           # by position
```

```python
#Create a Series for:
#Subjects: Physics, Chemistry, Maths

#Marks:78, 88, 92

#Print:
#The whole Series
#Only the Maths mark
subjects = ["Physics", "Chemistry", "Maths"] 
marks = [78, 88, 92] 
s = pd.Series(marks, index=subjects) 
print(s)
print(s["Maths"])
```
