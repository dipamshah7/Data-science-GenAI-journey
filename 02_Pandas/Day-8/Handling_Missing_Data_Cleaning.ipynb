```python
# Handling Missing Data & Cleaning
import pandas as pd
import numpy as np

data = {
    "Name": ["Dipam", "Raj", "Anita", "Karan", "Meena"],
    "Age": [28, np.nan, 27, 24, np.nan],
    "Marks": [90, 85, np.nan, 70, 95]
}

df = pd.DataFrame(data)
print(df)
```

```python
# Detect Missing Values
print(df.isnull()) #Shows True/False for missing

print(df.isnull().sum()) # To count missing values in each column
```

```python
# Drop Missing Values
print(df.dropna())
```

```python
# Fill Missing Values (Better Approach)

#Fill Age with mean:
df["Age"] = df["Age"].fillna(df["Age"].mean())


# Fill Marks with median:
df["Marks"] = df["Marks"].fillna(df["Marks"].median())
print(df)
```
