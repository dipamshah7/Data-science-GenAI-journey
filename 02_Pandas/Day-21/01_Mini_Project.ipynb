```python
import pandas as pd

data = {
    "Category": [
        "Electronics","Electronics","Electronics",
        "Electronics","Electronics","Electronics",
        "Furniture","Furniture","Furniture",
        "Furniture","Furniture","Furniture",
        "Clothing","Clothing","Clothing",
        "Clothing","Clothing","Clothing"
    ],

    "Product": [
        "Laptop","Laptop","Laptop",
        "Mobile","Mobile","Mobile",
        "Chair","Chair","Chair",
        "Table","Table","Table",
        "Shirt","Shirt","Shirt",
        "Jeans","Jeans","Jeans"
    ],

    "Month":[
        "Jan","Feb","Mar",
        "Jan","Feb","Mar",
        "Jan","Feb","Mar",
        "Jan","Feb","Mar",
        "Jan","Feb","Mar",
        "Jan","Feb","Mar"
    ],

    "Sales":[
        5000,6500,7000,
        3000,4200,4800,
        1500,1800,2000,
        2200,2500,2800,
        1200,1400,1600,
        1800,2100,2400
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Fix Month Order
month_order = ["Jan", "Feb", "Mar"]
df["Month"] = pd.Categorical(df["Month"], categories=month_order, ordered=True)

print(df)
```

```python
# Sort data by Category and Month
df = df.sort_values(by=["Category", "Month"])
print(df)
```

```python
# Product ranking within each category based on total product sales
df["Product_Total_Sales"] = (df.groupby(["Category","Product"])["Sales"].transform("sum"))
df["Product_Rank"] = df.groupby(["Category"])["Product_Total_Sales"].rank(method="dense",ascending=False)
print(df)
```

```python
# Category-wise Total sales 
df["Category_Total_Sales"] = df.groupby(["Category"])["Sales"].transform("sum")
print(df[["Category", "Product", "Month", "Sales", "Category_Total_Sales"]])
```

```python
# Contribution by %
df["Sales_Contribution"] = (df["Sales"] / df["Category_Total_Sales"]) * 100
print(df [["Category", "Product", "Month", "Sales", "Category_Total_Sales", "Sales_Contribution"]])
```

```python
# Cumulative sale within each category
df["Cumulative_Sales"] = df.groupby(["Category", "Product"])["Sales"].cumsum()
print(df [["Category", "Product", "Month", "Sales", "Category_Total_Sales", "Cumulative_Sales"]])
```

```python
# Previous month sales
df["Previous_Month_Sales"] = df.groupby(["Category", "Product"])["Sales"].shift(1)
print(df [["Category", "Product", "Month", "Sales","Previous_Month_Sales"]])
```

```python
# MoM Growth by using pct_change
df["MoM_Growth"] = df.groupby(["Category", "Product"])["Sales"].pct_change() * 100
print(df [["Category", "Product", "Month", "Sales","Previous_Month_Sales","MoM_Growth"]])
```

```python
# Identify high growth products
high_growth = df[df["MoM_Growth"] > 20]
print(high_growth[["Category", "Product", "Month", "Sales", "MoM_Growth"]])
```

```python
# Rolling average
df["Rolling_Avg"] = (df.groupby(["Category","Product"])["Sales"].transform(lambda x: x.rolling(2).mean()))
print(df [["Category", "Product", "Month", "Sales", "Rolling_Avg"]])
```

```python
# Show only these columns: | Category | Product | Month | Sales | Rank | Contribution_% | MoM_Growth_% | Rolling_Avg
# Sort by: Category → Rank
final_df = df[["Category", "Product", "Month", "Sales", "Product_Rank", "Sales_Contribution", "MoM_Growth", "Rolling_Avg"]]
final_df = final_df.sort_values(by=["Category", "Product_Rank"])
print(final_df)
```
