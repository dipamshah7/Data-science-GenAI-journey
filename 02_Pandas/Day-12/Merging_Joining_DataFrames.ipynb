```python
import pandas as pd

# Student basic info
students = pd.DataFrame({
    "Student_ID": [1, 2, 3, 4, 5],
    "Name": ["Dipam", "Raj", "Anita", "Karan", "Meena"],
    "Age": [28, 26, 27, 24, 26]
})

# Marks table
marks = pd.DataFrame({
    "Student_ID": [1, 2, 3, 4, 5],
    "Marks": [90, 85, 87, 70, 95]
})
```

```python
# This joins tables using common column.
df = pd.merge(students, marks, on="Student_ID")
print(df)

#✅ Inner Join (default) Only matching rows.

pd.merge(students, marks, on="Student_ID", how="inner")

#✅ Left Join All rows from left table, matching from right.
pd.merge(students, marks, on="Student_ID", how="left")
#✅ Right Join All rows from right table, matching from left.
pd.merge(students, marks, on="Student_ID", how="right")
#✅ Outer Join All rows from both tables, NaN for non-matching.
pd.merge(students, marks, on="Student_ID", how="outer")
```

```python
marks = pd.DataFrame({
    "Student_ID": [1, 2, 3, 4],
    "Marks": [90, 85, 87, 70]
})

df = pd.merge(students, marks, on="Student_ID", how="left")
print(df)
```

```python
#3️⃣ Concatenation (Stacking)

#Instead of joining by column, you can stack rows.

df1 = students.iloc[:3]
print(df1)
df2 = students.iloc[3:]

pd.concat([df1, df2])
```

```python
#Assignment
#1️⃣ Create new DataFrame called fees:
#Student_ID: [1,2,3,4]
#Fees_Paid: [True, True, False, True]
#2️⃣ Merge students + marks + fees using left join
#3️⃣ Show students who have NOT paid fees

Fees = pd.DataFrame({
    "Student_ID": [1, 2, 3, 4],
    "Fees_Paid": [True, True, False, True]
})

df = pd.merge(students, marks, on="Student_ID", how="left")
df = pd.merge(df, Fees, on="Student_ID", how="left")

print(df)

# Show students who have NOT paid fees
df[df["Fees_Paid"] == False].reset_index(drop=True)
```
