```python
import pandas as pd

data = [10, 20, 30, 40]
s = pd.Series(data)

print(s)
```

```python
s = pd.Series([90, 85, 88], index=["Math", "Science", "English"])
print(s)

print(s["Math"])      # by label
print(s[0])           # by position
```

```python
#Create a Series for:
#Subjects: Physics, Chemistry, Maths

#Marks:78, 88, 92

#Print:
#The whole Series
#Only the Maths mark
subjects = ["Physics", "Chemistry", "Maths"] 
marks = [78, 88, 92] 
s = pd.Series(marks, index=subjects) 
print(s)
print(s["Maths"])
```

```python
import pandas as pd

data = {
    "Name": ["Dipam", "Raj", "Anita"],
    "Age": [28, 25, 27],
    "Marks": [90, 85, 88]
}

df = pd.DataFrame(data)

print(df)
print(df["Marks"])
print(df.shape)

print(df.columns)
print(df.dtypes)
print(df.head())

print(df["Name"])
print(df[["Name", "Marks"]])

#df["Marks"]   → returns a Series

print(df[["Marks"]]) #→ returns a DataFrame


```

```python
print(df.loc[0])      # First row (by index label)
print(df.iloc[0])     # First row (by position)
```

```python
#Using the same dataframe:
#Print the second row using iloc
print(df.iloc[1])

#Print only the "Name" of third row using loc
print(df.loc[2, "Name"])
```

```python
#This filters rows where Marks > 85.
print(df[df["Marks"] > 85])
```

```python
#Using same dataframe:
#Print students whose Age is greater than 26
print(df[df["Age"] > 26])
```

```python
# Filter students whose:
# Marks > 85 AND Age < 28
print(df[(df["Marks"] > 85) & (df["Age"] < 28)])
```
