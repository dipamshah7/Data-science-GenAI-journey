```python
import pandas as pd

df = pd.read_csv("sales_data.csv")
print(df.head())
```

```python
print(df.info())
```

```python
#1️⃣ Check missing values
print(df.isnull().sum())
print(df)

#2️⃣ Fill missing Quantity with median
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].median())

#3️⃣ Fill missing Price with mean
df['Price'] = df['Price'].fillna(df['Price'].mean())

print(df)
```

```python
#4️⃣ Create new column Total_Sales = Quantity * Price
df['Total_Sales'] = df['Quantity'] * df['Price']
print(df)
```

```python
#5️⃣ Find total sales by City
sales_by_city = df.groupby('City')['Total_Sales'].sum().reset_index()
print(sales_by_city)
```

```python
#6️⃣ Find total sales by Product
sales_by_product = df.groupby('Product')['Total_Sales'].sum().reset_index()
print(sales_by_product)
```

```python
#7️⃣ Find city with highest revenue
max_city = sales_by_city.loc[sales_by_city['Total_Sales'].idxmax()]
print(max_city)
```

```python
import matplotlib.pyplot as plt

#Sales by City
plt.figure()
plt.bar(sales_by_city["City"], sales_by_city["Total_Sales"])
plt.xlabel("City")
plt.ylabel("Total Sales")
plt.title("Total Sales by City")
plt.show()
```

```python
#Sales by Product
plt.figure()
plt.bar(sales_by_product["Product"], sales_by_product["Total_Sales"])
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.title("Total Sales by Product")
plt.show()
```

```python
#Line Chart – Sales Trend by Date

#First calculate daily sales:
daily_sales = df.groupby("Order_Date")["Total_Sales"].sum().reset_index()

#Convert date properly
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

plt.figure()
plt.plot(daily_sales["Order_Date"], daily_sales["Total_Sales"])
plt.xlabel("Date")
plt.ylabel("Total Sales")
plt.title("Daily Sales Trend")
plt.xticks(rotation=45)
plt.show()
```

```python
# Pie Chart – Product Contribution
plt.figure()
plt.pie(
    sales_by_product["Total_Sales"],
    labels=sales_by_product["Product"],
    autopct="%1.1f%%"
)
plt.title("Sales Contribution by Product")
plt.show()
```
