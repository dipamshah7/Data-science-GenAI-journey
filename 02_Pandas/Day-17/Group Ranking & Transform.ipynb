```python
sale_data = {
    'Name': ['A', 'B', 'C', 'D'],
    'City': ['Mumbai', 'Mumbai', 'Delhi', 'Delhi'],
    'Revenue': [100, 200, 150, 300]}

import pandas as pd
df = pd.DataFrame(sale_data)
print(df)

#Rank customers within each city based on revenue
df['Rank'] = df.groupby('City')['Revenue'].rank(ascending=False)

#Transform()
df['total_city_revenue'] = df.groupby('City')['Revenue'].transform('sum')

df['contribution'] = df['Revenue'] / df['total_city_revenue']

#Get top 1 customer per city”
df['rank'] = df.groupby('City')['Revenue'].rank(ascending=False)
top_customers = df[df['rank'] == 1]

print(top_customers)

print(df)

```
