```python
import pandas as pd

data = {
    "Product": [
        "Laptop", "Mouse", "Keyboard", "Monitor", "Printer",
        "Tablet", "Headphones", "Camera", "Smartphone", "Speaker",
        "Laptop", "Mouse", "Keyboard", "Monitor", "Printer"
    ],
    "Category": [
        "Electronics", "Electronics", "Electronics", "Electronics", "Electronics",
        "Electronics", "Accessories", "Accessories", "Electronics", "Accessories",
        "Electronics", "Electronics", "Electronics", "Electronics", "Electronics"
    ],
    "Region": [
        "North", "North", "South", "South", "East",
        "East", "North", "South", "West", "West",
        "South", "East", "North", "West", "North"
    ],
    "Month": [
        "Jan", "Jan", "Jan", "Jan", "Jan",
        "Feb", "Feb", "Feb", "Feb", "Feb",
        "Mar", "Mar", "Mar", "Mar", "Mar"
    ],
    "Sales": [
        50000, 8000, 12000, 20000, 15000,
        30000, 7000, 25000, 60000, 9000,
        52000, 8500, 13000, 22000, 16000
    ]
}

df = pd.DataFrame(data)
```

```python
#Task 1 — Rank Products Within Category (by Sales)

df['Product_rank']= df.groupby('Category')['Sales'].rank(method='dense', ascending=False)

print(df[['Product', 'Category', 'Sales', 'Product_rank']])
```

```python
#Task 2 — Calculate Total Sales Per Category (Using transform)
df['Category_Total_Sales'] = df.groupby('Category')['Sales'].transform('sum')
print(df[['Product', 'Category', 'Sales', 'Category_Total_Sales']])
```

```python
#Task 3 — Contribution Percentage Round to 2 decimal places.
df['Contribution_Percentage'] = ((df['Sales'] / df['Category_Total_Sales']) * 100).round(2)
print(df[['Product','Sales', 'Category_Total_Sales', 'Contribution_Percentage']])
```

```python
#Task 4 — Top 2 Products Per Category
top_2_products = df[df['Product_rank'] <= 2]
print(top_2_products[['Category', 'Product', 'Sales', 'Product_rank']])
```
