```python
import pandas as pd

data = {
    "Name": ["Dipam", "Raj", "Anita", "Karan", "Meena"],
    "Age": [28, 26.33, 27, 24, 26.33],
    "Marks": [90, 85, 87.5, 70, 95]
}

df = pd.DataFrame(data)
```

```python
#🔹 1️⃣ Adding New Column
#Example: Add Pass/Fail column
#Condition:
#Marks ≥ 80 → Pass
#Marks < 80 → Fail

df["Result"] = df["Marks"] >= 80
print(df)
```

```python
#🔹 2️⃣ Conditional Column with Text
#Better version:

df["Result"] = df["Marks"].apply(lambda x: "Pass" if x >= 80 else "Fail")
print(df)
```

```python
#🔹 3️⃣ Creating New Calculated Column
#Example: Add 5 bonus marks

df["Final_Marks"] = df["Marks"] + 5
print(df)
```

```python
#🔹 4️⃣ Rename Columns
df.rename(columns={"Marks": "Score"}, inplace=True)
print(df)

#Modern way (recommended):
df = df.rename(columns={"Marks": "Score"})
print(df)
```

```python
#🔹 5️⃣ Drop Column

df = df.drop("Age", axis=1)

# axis=1 → column
# axis=0 → row
print(df)
```

```python
#🔹 6️⃣ Reset Index
#If you filtered rows and want clean index:

df = df.reset_index(drop=True)
```

```python
#Assignment:
#Using dataframe:

data = {
    "Name": ["Dipam", "Raj", "Anita", "Karan", "Meena"],
    "Age": [28, 26.33, 27, 24, 26.33],
    "Marks": [90, 85, 87.5, 70, 95]
}

df = pd.DataFrame(data)

#1️⃣ Create column "Grade": ≥ 90 → A, 80–89 → B, < 80 → C

df["Grade"] = df["Marks"].apply(lambda x: "A" if x >= 90 else ("B" if x >= 80 else "C"))

#2️⃣ Add column "Age_Group": Age ≥ 27 → "Senior", Age < 27 → "Junior"
df["Age_Group"] = df["Age"].apply(lambda x: "Senior" if x >= 27 else "Junior")

print(df)


#Alternative using np.select:
import numpy as np
conditions = [
    df["Marks"] >= 90,
    df["Marks"] >= 80
]

choices = ["A", "B"]

df["Grade"] = np.select(conditions, choices, default="C")
```
