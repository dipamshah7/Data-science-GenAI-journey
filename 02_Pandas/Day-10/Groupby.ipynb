```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Dipam", "Raj", "Anita", "Karan", "Meena"],
    "Age": [28, 26.33, 27, 24, 26.33],
    "Marks": [90, 85, 87.5, 70, 95]
}

df = pd.DataFrame(data)

df["Grade"] = np.select([df["Marks"] >= 90, df["Marks"] >= 80],["A", "B"],default="C")

df["Age_Group"] = df["Age"].apply(lambda x: "Senior" if x >= 27 else "Junior")

print(df)
```

```python
#🔹 1️⃣ Count Students Per Grade
print(df.groupby("Grade").size())
```

```python
#🔹 2️⃣ Average Marks Per Grade
print(df.groupby("Grade")["Marks"].mean())
```

```python
#🔹 3️⃣ Multiple Aggregations
print(df.groupby("Grade")["Marks"].agg(["mean", "max", "min", "count"]))
```

```python
#🔹 4️⃣ Group by Age_Group and Calculate Average Marks
print(df.groupby("Age_Group")["Marks"].mean())
```

```python
#🔹 5️⃣ Reset Index (Cleaner Output)
result = df.groupby("Grade")["Marks"].mean().reset_index()
print(result)
```

```python
#🧪 Your Task (Mini Analysis)

#Using current dataframe:

#1️⃣ Find average marks for Senior and Junior
print(df.groupby("Age_Group")["Marks"].mean().reset_index())

#2️⃣ Find count of students in each Grade
print(df.groupby("Grade").size().reset_index(name="Count"))

#3️⃣ Find max marks in each Grade
print(df.groupby("Grade")["Marks"].max().reset_index())
```
