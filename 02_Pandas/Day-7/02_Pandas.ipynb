```python
import pandas as pd

data = {
    "Name": ["Dipam", "Raj", "Anita", "Karan", "Meena"],
    "Age": [28, 25, 27, 24, 29],
    "Marks": [90, 85, 88, 70, 95]
}

df = pd.DataFrame(data)
```

```python
print(df.describe())
```

```python
# Basic Column Statistics
print(df["Marks"].mean())
print(df["Marks"].max())
print(df["Marks"].min())
print(df["Marks"].sum())
```

```python
#Counts frequency of values
print(df["Age"].value_counts())
```

```python
#Sort by Marks ascending
print(df.sort_values("Marks"))

#Sort by Marks descending
print(df.sort_values("Marks", ascending=False))
```

```python
print(df.describe())
print(df["Marks"].mean())
print(df.sort_values("Age", ascending=False))
```

```python
#If you want to permanently sort:

df = df.sort_values("Age", ascending=False)
```
