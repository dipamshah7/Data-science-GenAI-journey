```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Dipam", "Raj", "Anita", "Karan", "Meena"],
    "Age": [28, 26.33, 27, 24, 26.33],
    "Marks": [90, 85, 87.5, 70, 95]
}

df = pd.DataFrame(data)

df["Grade"] = np.select(
    [df["Marks"] >= 90, df["Marks"] >= 80],
    ["A", "B"],
    default="C"
)

df["Age_Group"] = df["Age"].apply(lambda x: "Senior" if x >= 27 else "Junior")

print(df)
```

```python
#🔹 1️⃣ Group by Multiple Columns

# Example: Count students by Grade AND Age_Group

print(df.groupby(["Grade", "Age_Group"]).size())
```

```python
#🔹 2️⃣ Sort Grouped Results

# Example: Highest average marks per Grade first

result = df.groupby("Grade")["Marks"].mean().reset_index()
print(result.sort_values("Marks", ascending=False))
```

```python
#🔹 3️⃣ Pivot Table (Very Important 🔥)

#Pivot table = Excel-style summary table

#Example:

#Average Marks by Grade & Age_Group

pivot = df.pivot_table(
    values="Marks",
    index="Grade",
    columns="Age_Group",
    aggfunc="mean"
)

print(pivot)
```

```python
# Pivot Table with Multiple Aggregations

pivot = df.pivot_table(
    values="Marks",
    index="Grade",
    aggfunc=["mean", "max", "count"]
)

print(pivot)
```

```python
#assignment

#1️⃣ Create pivot table showing count of students by Grade and Age_Group

pivot = df.pivot_table(
    values="Name",
    index="Grade",
    columns="Age_Group",
    aggfunc=["count"]
)
print(pivot)

#2️⃣ Create pivot table showing max marks by Age_Group
pivot = df.pivot_table(
    values="Marks",
    columns="Age_Group",
    aggfunc="max"
)
print(pivot)

#3️⃣ Sort pivot table by Grade descending
pivot = df.pivot_table(
    values="Marks",
    index="Grade",
    aggfunc="mean"
).sort_values("Marks", ascending=False)
print(pivot)
```
